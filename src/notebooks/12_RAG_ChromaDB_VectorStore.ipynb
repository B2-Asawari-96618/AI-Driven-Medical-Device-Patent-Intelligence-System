```python
import pandas as pd
import numpy as np

# Load chunk dataframe
chunk_df = pd.read_csv("../data/processed/patent_chunks.csv")

# Load chunk embeddings
chunk_embeddings = np.load("../data/processed/chunk_embeddings.npy")

print(chunk_df.shape)
print(chunk_embeddings.shape)
```

```python
#Step 1 Create Database
import chromadb


client = chromadb.PersistentClient(
path="../vector_db"
)
```

```python
#Step 2 Create Collection
collection=client.get_or_create_collection(

name="medical_patent_chunks"

)
```

```python
from tqdm import tqdm

BATCH_SIZE = 5000

for start in tqdm(range(0, len(chunk_df), BATCH_SIZE)):
    
    end = start + BATCH_SIZE

    collection.add(
        ids=chunk_df["chunk_id"].iloc[start:end].astype(str).tolist(),

        documents=chunk_df["chunk_text"].iloc[start:end].tolist(),

        embeddings=chunk_embeddings[start:end].tolist(),

        metadatas=chunk_df[
            ["patent_id", "title", "country", "assignee", "year"]
        ].iloc[start:end].to_dict("records")
    )

print(" All chunks inserted successfully!")
```
