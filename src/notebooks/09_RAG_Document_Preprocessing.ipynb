```python
#Step 1 Import Libraries
import pandas as pd
import numpy as np
```

```python
#Step 2 Load Dataset
DATA_PATH = "../data/processed/cleaned_patent_data.csv"

df = pd.read_csv(DATA_PATH)

df.shape
```

```python
#Step 3 Check Missing Values
df[
[
"title",
"abstract_en",
"claims_clean",
"description"
]
].isnull().sum()
```

```python
#Step 4 Select Text Fields
df["rag_text"] = (
    
    "Title: "
    + df["title_en"].fillna("")
    
    + "\n\nAbstract: "
    + df["abstract_en"].fillna("")
    
    + "\n\nClaims: "
    + df["claims_clean"].fillna("")
    
    + "\n\nDescription: "
    + df["description"].fillna("")
)
```

```python
#Step 5 Add Metadata
rag_df = df[
[
"patent_id",
"title_en",
"country",
"assignee_en",
"filing_year",
"ipc_codes",
"cpc_codes",
"forward_citation_count",
"backward_citation_count",
"legal_status"
]
].copy()
```

```python
#add
rag_df["text"] = df["rag_text"]
```

```python
#Step 6 Remove Empty Documents
rag_df = rag_df[
rag_df["text"].str.len()>100
]
```

```python
#Step 7 Save
rag_df.to_csv(
"../data/processed/rag_documents.csv",
index=False
)
```
