Notebook 14: Text Chunking
Objective
Split large patent documents into smaller searchable pieces.
Why?
Because:
Patent = 20,000 characters

LLM cannot process all patents together
Instead:
Patent

 |
 |
Chunk 1
Chunk 2
Chunk 3
Chunk 4

```python
import pandas as pd
```

```python
#Step 1 Load Documents
df = pd.read_csv(
"../data/processed/rag_documents.csv"
)
```

```python
#Step 2 Import Splitter
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
#Step 3 Create Chunker
#Recommended for patents:
splitter = RecursiveCharacterTextSplitter(

    chunk_size=1000,

    chunk_overlap=200,

    length_function=len

)
```

Why?
1000 characters gives meaningful context
200 overlap prevents information loss

```python
#Step 4 Create Chunks
chunks=[]


for _,row in df.iterrows():

    documents = splitter.split_text(
        row["text"]
    )


    for i,doc in enumerate(documents):

        chunks.append(
        {

        "chunk_id":
        f"{row['patent_id']}_{i}",

        "patent_id":
        row["patent_id"],

        "chunk_text":
        doc,

        "title":
        row["title_en"],

        "country":
        row["country"],

        "assignee":
        row["assignee_en"],

        "year":
        row["filing_year"]

        }
        )
```

```python
#convert
chunk_df=pd.DataFrame(chunks)

chunk_df.shape
```

```python
#save
chunk_df.to_csv(
"../data/processed/patent_chunks.csv",
index=False
)
```
